### Unknown row per dimension

```python
run_sql("""
    select 'dim_pos' as dim, count(*) as unknown_rows from marts.dim_pos where pos_key = '-1'
    union all select 'dim_campaign', count(*) from marts.dim_campaign where campaign_key = -1
    union all select 'dim_route', count(*) from marts.dim_route where route_key = -1
    union all select 'dim_date', count(*) from marts.dim_date where date_key = -1
    union all select 'dim_visit_status', count(*) from marts.dim_visit_status where status_key = -1
""")
```
