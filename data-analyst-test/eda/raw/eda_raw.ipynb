### Setup

```python
import unicodedata
from pathlib import Path

import pandas as pd

RAW_DIR = Path("../../data/raw")
csv_paths = sorted(RAW_DIR.glob("*.csv"))
raw_tables = {path.stem: pd.read_csv(path) for path in csv_paths}
list(raw_tables.keys())
```

```python
def summarize_table(df: pd.DataFrame, pk: str) -> None:
    """Run the standard eda checks for one raw table."""
    df.info()
    display(df.head())

    null_counts = df.isna().sum()
    print("nulls per column:")
    print(null_counts[null_counts > 0])

    dup_pk = df[pk].duplicated().sum()
    print(f"pk={pk} unique={df[pk].nunique()} rows={len(df)} dup_pk={dup_pk}")

    str_cols = df.select_dtypes(include="str").columns
    print("distinct values per text column:")
    print({col: df[col].nunique() for col in str_cols})
```

### Repeated keys and rows

```python
PKS = {
    "raw_clients": "client_id", "raw_projects": "project_id", "raw_campaigns": "campaign_id",
    "raw_questions": "question_id", "raw_workers": "employee_id", "raw_pos": "intervention_point_id",
    "raw_routes": "route_id", "raw_route_employee": "route_employee_id",
    "raw_visits": "visit_id", "raw_responses": "answer_id",
}
rows = [(name, raw_tables[name][key].duplicated().sum(), raw_tables[name].duplicated().sum()) for name, key in PKS.items()]
pd.DataFrame(rows, columns=["table", "repeated ids", "repeated rows"])
```

### Dates

```python
def read_date(col):
    iso = pd.to_datetime(col, format="%Y-%m-%d", errors="coerce")
    return iso.fillna(pd.to_datetime(col, format="%d/%m/%Y", errors="coerce"))


c_start = read_date(campaigns["campaign_start_date"])
c_end = read_date(campaigns["campaign_end_date"])
v_date = read_date(visits["visit_date"])
print("campaign dates not read:", (campaigns["campaign_start_date"].notna() & c_start.isna()).sum() + (campaigns["campaign_end_date"].notna() & c_end.isna()).sum())
print("visit dates not read:", (visits["visit_date"].notna() & v_date.isna()).sum())
print("visit dates from", v_date.min().date(), "to", v_date.max().date())
print("campaigns ending before start:", (c_end < c_start).sum())
display(campaigns.loc[c_end < c_start, ["campaign_id", "campaign_start_date", "campaign_end_date"]])
```

### Visits outside campaign dates

```python
dates = campaigns[["campaign_id"]].assign(start=c_start, end=c_end)
check = visits.assign(day=v_date).merge(dates, on="campaign_id")
print("visits outside campaign dates:", (check["day"].notna() & ((check["day"] < check["start"]) | (check["day"] > check["end"]))).sum())
```

### Province spelling

```python
def plain(text):
    return unicodedata.normalize("NFKD", text).encode("ascii", "ignore").decode().upper().strip()


rows = []
for name, df, col in [("workers", workers, "employee_address_province"), ("points of sale", pos, "intervention_point_province")]:
    values = df[col].dropna()
    rows.append((name, values.nunique(), values.map(plain).nunique(), values.str.contains(r", (?:La|Las|El|Los)$", case=False).sum()))
pd.DataFrame(rows, columns=["table", "spellings", "after upper case and no accents", "like Rioja, La"])
```

### Spaces and empty text

```python
rows = []
for name, df in raw_tables.items():
    text = df.select_dtypes(include="str")
    edge = double = empty = 0
    for col in text:
        values = text[col].dropna()
        edge += (values != values.str.strip()).sum()
        double += values.str.contains(r"\s{2,}").sum()
        empty += (values.str.strip() == "").sum()
    rows.append((name, edge, double, empty))
pd.DataFrame(rows, columns=["table", "edge spaces", "double spaces", "empty text"])
```
