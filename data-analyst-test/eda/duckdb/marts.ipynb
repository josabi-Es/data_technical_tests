### Null audit

Marts keep the nulls of the source. Joins are LEFT and do not create nulls.

```python
# raw vs mart nulls
run_sql("""
    select 'responses.question_id' as col,
           (select count(*) filter (where question_id is null) from raw.raw_responses) as raw_nulls,
           (select count(*) filter (where question_id is null) from marts.fct_responses) as mart_nulls
    union all
    select 'responses.answer',
           (select count(*) filter (where answer is null) from raw.raw_responses),
           (select count(*) filter (where answer is null) from marts.fct_responses)
    union all
    select 'responses.question_type',
           (select count(*) filter (where question_type is null) from raw.raw_responses),
           (select count(*) filter (where question_type is null) from marts.fct_responses)
    union all
    select 'visits.visit_type',
           (select count(*) filter (where visit_type is null) from raw.raw_visits),
           (select count(*) filter (where visit_type is null) from marts.fct_visits)
    union all
    select 'visits.visit_time',
           (select count(*) filter (where visit_time is null) from raw.raw_visits),
           (select count(*) filter (where visit_time is null) from marts.fct_visits)
    union all
    select 'visits.is_client_billable',
           (select count(*) filter (where is_client_billable is null) from raw.raw_visits),
           (select count(*) filter (where is_client_billable is null) from marts.fct_visits)
    union all
    select 'campaign.visit_duration_minutes',
           (select count(*) filter (where visit_duration_minutes is null) from raw.raw_campaigns),
           (select count(*) filter (where visit_duration_minutes is null) from marts.dim_campaign)
""")
```

```python
# join orphans
run_sql("""
    select
        count(*) filter (where dc.campaign_id is null) as no_campaign,
        count(*) filter (where dp.intervention_point_id is null) as no_pos,
        count(*) filter (where fr.question_id is not null and dq.question_id is null) as orphan_question,
        count(*) filter (where fr.question_id is null) as answers_without_question
    from marts.fct_responses as fr
    left join marts.dim_campaign as dc on fr.campaign_id = dc.campaign_id
    left join marts.dim_pos as dp on fr.intervention_point_id = dp.intervention_point_id
    left join marts.dim_question as dq on fr.question_id = dq.question_id
""")
```

```python
# worker null reasons
run_sql("""
    select count(*) filter (where employee_id is null) as no_worker,
           count(*) filter (where employee_id is null and route_id = -1) as no_route,
           count(*) filter (where employee_id is null and route_id <> -1) as route_without_main
    from marts.fct_visits
""")
```
