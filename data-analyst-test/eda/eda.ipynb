## summarize_table

One function, same checks for every table below, so results are easy to
compare across the 10 raw csvs.

```python
def summarize_table(df: pd.DataFrame, pk: str) -> None:
    """Run the standard eda checks for one raw table."""
    df.info()
    display(df.head())

    null_counts = df.isna().sum()
    print("nulls per column:")
    print(null_counts[null_counts > 0])

    dup_pk = df[pk].duplicated().sum()
    print(f"pk={pk} unique={df[pk].nunique()} rows={len(df)} dup_pk={dup_pk}")

    str_cols = df.select_dtypes(include="str").columns
    print("distinct values per text column:")
    print({col: df[col].nunique() for col in str_cols})
```

## Data quality issues

At least three issues found during the EDA above, with real evidence and
the proposed fix:

1. **Duplicate primary keys**: `raw_clients.client_id`, `raw_workers.employee_id`
   and `raw_visits.visit_id` all have exact duplicate rows (e.g. visit_id 20
   appears twice, same data). Staging must deduplicate with a distinct or a
   row_number over the natural key before using it as a join key.
2. **Inconsistent country and province casing**: `raw_clients.country` has
   `spain`, `SPAIN` and `España` for the same value, just casing, verified
   at byte level, not an encoding bug. `raw_pos.intervention_point_province`
   mixes case and includes a non spanish province (`Évora`, Portugal). Fix
   with trim and uppercase, same pattern as `clean_string` in the reference
   dbt project.
3. **Mixed date formats**: `raw_campaigns.campaign_start_date` has 12 rows in
   `dd/mm/yyyy` format while the rest are `yyyy-mm-dd`. Staging must parse
   both formats explicitly before casting to date, otherwise pandas or dbt
   will silently misread day and month.
4. **Soft deletes not filtered**: `raw_route_employee` has a `deleted_at`
   column, rows are not physically removed. Staging must filter
   `deleted_at is null` unless historical rows are needed on purpose.
5. **Unexpected column not documented**: `raw_responses` has an
   `expected_answer` column not mentioned in the README, worth checking if
   it should feed a correctness metric on worker answers.

## Star schema proposal

Grain and dimensions proposed from the data model:

- **dim_client**, **dim_project**, **dim_campaign**: one row per entity,
  campaign carries the date range and planned visits.
- **dim_pos**: one row per intervention point (province, location).
- **dim_worker**: one row per employee.
- **dim_question**: one row per question, campaign and category.
- **dim_route**: one row per route, campaign and dates.
- **fct_visits**: grain is one row per visit. Foreign keys to campaign, pos,
  route. Measures like visit status and planned versus done.
- **fct_responses**: grain is one row per answer. Foreign keys to visit and
  question, this is the table that feeds the dashboard KPIs on completion
  and answer quality.

Two fact tables because visits and responses have a different grain, one
visit can have many responses.
